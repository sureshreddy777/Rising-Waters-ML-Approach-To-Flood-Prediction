```python
x = dataset.iloc[:,2:7].values
y = dataset.iloc[:,10:].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=10)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.fit_transform(x_test)
```

```python
from joblib import dump
```

```python
dump(sc,"transform.save")
```

```python
from sklearn import tree,ensemble,neighbors
import xgboost
```

```python
dtree = tree.DecisionTreeClassifier()
rf = ensemble.RandomForestClassifier()
knn = neighbors.KNeighborsClassifier()
xgb = xgboost.XGBClassifier()
```

```python
dtree.fit(x_train,y_train)
rf.fit(x_train,y_train)
knn.fit(x_train,y_train)
xgb.fit(x_train,y_train)
```

```python
from sklearn import metrics
```

```python
dtree_pred = dtree.predict(x_test)
rf_pred = rf.predict(x_test)
knn_pred = knn.predict(x_test)
xgb_pred = xgb.predict(x_test)

print(f"Decision Tree Accuracy: {metrics.accuracy_score(y_test, dtree_pred)}")
print(f"Random Forest Accuracy: {metrics.accuracy_score(y_test, rf_pred)}")
print(f"K-Nearest Neighbors Accuracy: {metrics.accuracy_score(y_test, knn_pred)}")
print(f"XGBoost Accuracy: {metrics.accuracy_score(y_test, xgb_pred)}")
```

```python
metrics.confusion_matrix(y_test,xgb_pred)
```

```python
print(metrics.accuracy_score(y_test,xgb_pred))
```

```python
print(metrics.precision_score(y_test,xgb_pred))
```

```python
print(metrics.recall_score(y_test,xgb_pred))
```
